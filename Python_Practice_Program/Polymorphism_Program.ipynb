```python
# POLYMORPHISM
# 1. Function Polymorphism  (Polymorphism using bulit-in-function)
print(len("Python"))     #String length
print(len([1,2,3,4]))    #List length
print(len((10,20)))      #Tuple length
```

```python
# 2. Metthod Overriding(Method Overriding is a feature inOOP where a child class provides a different implementation of a method that already exists in the parent)
class Employee:
  def work(self):
    print("Employee is woring")
class Developer(Employee):
  def work(self):
    print("Developer is writing code")
class Tester(Employee):
  def work(self):
    print("Testeer is testing application")
d=Developer()
t=Tester()

d.work()
t.work()
```

```python
# 3. Duck Typing
class Dog:
  def sound(self):
    print("Dog sound")
class Cat:
  def sound(self):
    print("Cat meows")
class Bird:
  def sound(self):
    print("Bird chirps")
def animal_sound(animal):
  animal.sound()
d=Dog()
c=Cat()
b=Bird()
animal_sound(d)
animal_sound(c)
animal_sound(b)
```

```python
# 4. Operator Overloading
print(10+20)           #Addition
print("Hello"+"AI")    #String concatenation
print([1,2]+[3,4])     #List merging
```

```python
# 5. Runtime Polymorphism (Industry-Level)
class Payment:
  def process_payment(self):
    print("Processing payment")
class UPI(Payment):
  def process_payment(self):
    print("Processing UPI payment")
class CreditCard(Payment):
  def process_payment(self):
    print("Processing Credit Card payment")
class Wallet(Payment):
  def process_payment(self):
    print("Processing Wallet payment")
payments=[UPI(),CreditCard(),Wallet()]
for p in payments:
  p.process_payment()
```
