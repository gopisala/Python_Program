```python
#PYTHON-SET DATA STRUTURE
#Creating same type of elements by using set
s={10,20,30,40}
print(s)
print(type(s))
```

```python
#Creating different type of elements by using
s={10,"Daniel",30.9,"Prasad",40}
print(s)
print(type(s))
```

```python
#Creating set by using range(p)
s=set(range(5))
print(s)
```

```python
#set not allowed duplicates
s={10,20,30,40,10,10}
print(s)
print(type(s))
```

```python
#creating set by using se(paraeter1)function
r=range(0,10)
l=set(r)
print(l)
```

```python
#Empty curly braces is not set,it considers as a dictionary
s=()
print(s)
print(type(s))
```

```python
#Using set( function to create empty set)
s=set()
print(s)
print(type(s))
```

```python
#Len(p) function

#to find length of set
n={10,20,30,40,50}
print(len(n))
```

```python
#To find length of set
n={10,20,30,40,50,10,10,10}
print(len(n))
```

```python
#Printing set data structure methods by using dir(set) function
print(dir(set))
```

```python
#add(p) function can add element to the element to the set
s={10,20,30}
s.add(40)
print(s)
```

```python
#REMOVE(P) METHOD
#Remove(p) method
s={40,10,30,20}
s.remove(30)
print(s)
```

```python
#CLEAAR() METHOD
# clear method in set
s={10,20,30}
print(s)
s.clear()
print(s)
```

```python
#MEMBERSHIP OPERATORS:(IN AND NOT IN)
# in operator
s={1,2,3,"daniel"}
print(s)
print(1 in s)
print("z" in s)
```

```python
#Set comprehension
s={x*x for x in range(5)}
print(s)
```

```python
#Remove duplicates from list
a={10,20,30,40,10,20,30}
s=set(a)
print(s)
```
