```python
#String basics & indexing(string indexing-Positive and Negative)
s="PYTHON"
#Positive indexing
print("Index 0:",s[0])  #P
print("Index 2:",s[2])  #T
print("Index 5:",s[5])  #n
#negative indexing
print("Index -1:",s[-1])  #N(last character)
print("Index -3:",s[-3])  #H
print("Index -6:",s[-6])  #p(first character)
```

```python
#Find character at given index and index of a character
s="banana"
#character at index
print("char at index 2:",s[2])   #n
#index of first occurrence
print("First index of a:",s.index("a")) #1
#All positions of a character
positions=[i for i, ch in enumerate(s) if ch=="a"]
print("All positions of a:",positions)   #[1,2,5]
```

```python
# Immutability of strings
s="helo"
# this would cause typeerror:
#s[0]="H"  <--Not allowed
#correct way:build a new string
modified="H"+s[1:]
print("Original:",s)     #hello
print("Modified:",modified)   #Hello
#using replace()
new_s=s.replace("h","H")
print("replace():",new_s)    #hello
```

```python
#Check Each Character-vowel or consonant
s="Python"
vowels="aeiouAEIOU"
for i,ch in enumerate(s):
 kind="Vowel"  if ch in vowels else "Consonant"
print(f"Index {i}:{ch}->{kind}")
```

```python
#SLICING
#string slicing-all forms
s="ABCDEFGHIJ"
print(s[2:7])
print(s[:4])
print(s[6:])
print(s[::2])
print(s[::-1])
print(s[1:8:3])
print(s[-4:-1])
```

```python
#reverse a string using slicing
s="Data Science"
#Method 1:slicing(Most pythonic)
rev1=s[::-1]
print("Slicing :",rev1)
#Method 2: reverrsed()+join
rev2="".join(reversed(s))
print("reversed():",rev2)
#check palindrome
word="racecar"
print(f"Is {word!r} palindrome?", word==word[::-1])

```

```python
#Extract words using slicing and split()
sentence="Machine Learning is the future"
words=sentence.split()
print("First word:",words[0])
print("Last word:",words[-1])
print("Middle words:",words[1:-1])

#character slicing
print("First 7 chars:",sentence[:7])
print("Last 6 chars:",sentence[-6:])
```

```python
#STRING ADDITION
#string Concantenation-all methods
first="Data"
second="Science"
#1. +operator
print(first +""+ second)
#2. +=operator
result=first
result+=""+second
print(result)
#3. f-string (fastes for few variables)
print(f"{first}{second}")
#4.join()-best for many strings
words=["Python","pands","ML","NLP"]
print(",".join(words))
#5.% formating(legacy)
print("%s %s" % (first,second))
```

```python
#Concatenate String with Numbers Using str()
name="Score"
value=95
#Wrong--raises TypeError:
#print(name+value)
#Fix 1:explicit str()
print(name+":"+str(value))
#Fix 2:f-string(cleanest)
print(f"{name}:{value}")
#Fix3:format()
print("{}:{}".format(name,value))
#practical example
for subj,marks in [("Python",88),("ML",92),("NLP",85)]:
 print(f"{subj:<8}:{marks}/100")
```

```python
#Draw paaters using String Multiplication
#Progress bar
def progress_bar(pct,width=20):
  filled=int(width*pct/100)
  bar="#"*filled+"-"*(width-filled)
  return f"[{bar}]{pct}%"
  for p in [0,25,50,75,100]:
    print(progress_bar(p))
#star rating
def stars(n,max_n=5):
  return"*"*n+"."*(max_n-n)
  print("Rating:",stars(4))
```

```python
#strip(),istrip(),rstrip()-whitespace removal
s=" ***Hello World***"
print(repr(s.strip()))      #both ends
print(repr(s.lstrip()))     #left only
print(repr(s.rstrip()))      #right only

#Remove specific characters
print(repr(s.strip("*")))    #removes spaces AND*

#Practical:clean CSV data
csv_row="Alice,25,chennai"
clean=[x.strip()for x in csv_row.split(",")]
print(clean)
```
