```python
# 6. One Parameterised constructor
class Employee:
  def __init__(self,number):
    self.number=number
    print("Employee id is:",self.number)
e1=Employee(1)
e2=Employee(2)
e3=Employee(3)
```

```python
# 7.One parameterised constructur and instance method
class Employee:
  def __init__(self,number):
    self.number=number
  def display(self):
    print("Employee id is:",self.number)
e1=Employee(1)
e2=Employee(2)
e3=Employee(3)
e1.display()
e2.display()
e3.display()
```

```python
# 8.Two parameterised constructor and instance method
class Employee:
  def __init__(self,number,name):
    self.number=number
    self.name=name
  def display(self):
    print("Hello my id is:",self.number)
    print("My name is:",self.name)
e1=Employee(1,"Raja")
e1.display()
e2=Employee(2,"gopika")
e2.display()
```

```python
# 9. Three parameterised constructor and instance method
class Employee:
  def __init__(self,number,name,age):
    self.number=number
    self.name=name
    self.age=age
  def display(self):
    print("Hello my id is:",self.number)
    print("My name is;",self.name)
    print("my age is:",self.age)
e1=Employee(1,"Raja",22)
e1.display()
e2=Employee(2,"Gopika",23)
e2.display()
e3=Employee(3,"Sumathi",45)
e3.display()
```

```python
# 10. instance variable
class Employee:
  def __init__(self,name,number):
     self.name=name
     self.number=number
e1=Employee("Raja",101)
e2=Employee("Gopika",102)
print("Employee 1 info:")
print("name:",e1.name)
print("Id:",e1.number)
print("Employee2  info:")
print("name:",e2.name)
print("id:",e2.number)
```

```python
# 11.Initializing instance variable inside Constructor
class Employee:
  def __init__(self):
    self.name="Raja"
    self.number="638"
    self.age="22"
emp=Employee()
print("Employee name:",emp.name)
print("Employee number:",emp.number)
print("Employee age:",emp.age)
```
