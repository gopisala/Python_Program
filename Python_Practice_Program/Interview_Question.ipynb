```python
# Interrview question
# 1.Reverse a String
# write a Python program to revesea given string without using the bulit-in reverse() function.
def reverse_string(s):
  result=""
  for ch in s:
    result=ch+result
  return result
print(reverse_string('hello'))
```

```python
# 2.Check if a number is prime
#Write a Python program to check whether a given number is a prime number
def is_prime(n):
  if n<2:
    return False
  for i in range(2,int(n**0.5)+1):
    if n%i==0:
      return False
  return True
num=29
num1=15
print('Prime'if is_prime(num)else'Not Prime')
print('Prime'if is_prime(num1)else'Not Prime')
```

```python
# Generate The Fibonacci Series
def fibonacci(n):
  series=[]
  a,b=0,1
  for c in range(n):
    series.append(a)
    a,b=b,a+b
  return series
print(fibonacci(4))

```

```python
#4.Factorial of a number(recursion)
def factorial(n):
  if n==0 or n==1:
    return 1
  return n* factorial(n-1)
print(factorial(5))
```

```python
#Check Palindrome String
def is_palindrome(s):
  s=s.lower().replace('',"")
  return s==s[::-1]
print(is_palindrome('Madam'))
print(is_palindrome('Dad'))
print(is_palindrome('A man a plan a canal Panama'))
```

```python
from os import cpu_count
# 8.Count Vowels and Consonants in a String
def count_vowels_consonants(s):
  vowels='aeiouAEIOU'
  v_count=c_count=0
  for ch in s:
    if ch.isalpha():
      if ch in vowels:
        v_count+=1
      else:
        c_count+=1
  return v_count,c_count
v,c=count_vowels_consonants('Interview')
print(f'Vowels={v},Consonants={c}')

```

```python
# 9. Check Armstrong Number
def is_armstrong(n):
  digits=str(n)
  power=len(digits)
  total=sum(int(d)**power for d in digits)
  return total==n
num=153
print('Armstrong Number'if is_armstrong(num) else 'Not an Armstrong Number')
```

```python
# 10. Remove duplicates from a List
def remove_duplicates(lst):
  seen=set()
  result=[]
  for item in lst:
    if  item not in seen:
      seen.add(item)
      result.append(item)
  return result
print(remove_duplicates([1,2,2,3,4,4,5]))


```

```python
# 11. Find the Second Largest Number in a List
def second_largest(lst):
  unique=list(set(lst))
  unique.sort(reverse=True)
  if len(unique)<2:
    return None
  return unique[1]
print(second_largest([10,25,3,47,47,8]))
```

```python
#12.Check if Two Strings are Anagrams
def is_anagram(s1,s2):
  s1=s1.replace('',"").lower()
  s2=s2.replace('',"").lower()
  return sorted(s1)==sorted(s2)
print(is_anagram('listen','silent'))
print(is_anagram('hello','world'))
```

```python
# 13.FizzBuzz
def fizzbuzz(n):
  result=[]
  for i in range(1,n+1):
    if i % 15==0:
      result.append('FizzBuzz')
    elif i % 3==0:
      result.append('Fizz')
    elif i % 5==0:
      result.append('Buzz')
    else:
      result.append(str(i))
  return result
print(''.join(fizzbuzz(15)))
```

```python
# 14. Find the GCD(HCF) of Two numbers
def gcd(a,b):
  while b:
    a,b=b,a%b
  return a
print(gcd(48,18))
import math
print(math.gcd(48,18))
```

```python
# 15.Count the Frequency of Each Element in a List
def count_frequency(lst):
  freq={}
  for item in lst:
    freq[item]=freq.get(item,0)+1
  return freq
print(count_frequency(['a','b','a','c','b','a']))
from collections import Counter
print(Counter(['a','b','a','c','b','a']))


```
