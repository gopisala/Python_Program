```python
#Python Inheritance & Abstraction
# 1. Single Inheritance
class Person:
  def __init__(self,name,age):
    self.name=name
    self.age=age
  def show_details(self):
    print(f"Name:{self.name},Age:{self.age}")
class Student(Person):
  def __init__(self,name,age,roll_no,course):
    super().__init__(name,age)
    self.roll_no=roll_no
    self.course=course
  def show_details(self):
    super().show_details()
    print(f"Roll No:{self.roll_no},Course:{self.course}")
s1=Student("Anjali Rao",20,"SC101","Computer Science")
s1.show_details()
```

```python
# 2. Multilevel Inheritance
class Person:
  def __init__(self,name):
    self.name=name
  def show_person(self):
    print(f"Name:{self.name}")
class Employee(Person):
  def __init__(self,name,emp_id):
    super().__init__(name)
    self.emp_id=emp_id
  def show_employee(self):
    self.show_person()
    print(f"Employee ID:{self.emp_id}")
class Manager(Employee):
  def __init__(self,name,emp_id,team_size):
    super().__init__(name,emp_id)
    self.team_size=team_size
  def show_manager(self):
    self.show_employee()
    print(f"Team Size:{self.team_size}")
m1=Manager("Vikram Singh","E501",12)
m1.show_manager()
```

```python
# 3. Multiple Inheritance
class Flyable:
  def fly(self):
    print("I can fly high in the sky.")
class Swimmable:
  def swim(self):
    print("i can swim in the water.")
class Duck(Flyable,Swimmable):
  def __init__(self,name):
    self.name=name
  def introduce(self):
    print(f" I am {self.name},a duck.")
donaid=Duck("Donald")
donaid.introduce()
donaid.fly()
donaid.swim()
```

```python
# 4. Hierarchical Inheritance
class Shape:
  def __init__(self,name):
    self.name=name
  def describe(self):
    print(f"This is a {self.name}.")
class Circle(Shape):
  def __init__(self,radius):
    super().__init__("Circle")
    self.radius=radius
  def area(self):
    return 3.14159*self.radius**2
class Rectangle(Shape):
  def __init__(self,length,width):
    super().__init__("Rectangle")
    self.length=length
    self.width=width
  def area(self):
    return self.length*self.width
class Triangle(Shape):
  def __init__(self,base,height):
    super().__init__("Triangle")
    self.base=base
    self.height=height
  def area(self):
    return 0.5* self.base*self.height
shapes=[Circle(4),Rectangle(5,6),Triangle(6,8)]
for shape in shapes:
  shape.describe()
  print(f"Area:{shape.area():.2f}\n")
```

```python
# 5.Method Overriding with super()
class Vehicle:
  def __init__(self,brand):
    self.brand=brand
  def start(self):
    print(f"{self.brand} vehicle is starting in the standard way.")
class Car(Vehicle):
  def start(self):
    super().start()
    print(f"{self.brand} car engine roars to life a push-button start.")
car=Car("Tesla")
car.start()
```

```python
# 6.Abstract Class and Abstract Method
from abc import ABC,abstractmethod
class Payment(ABC):
  @abstractmethod
  def pay(self,amount):
    pass
class CreditCardPayment(Payment):
  def pay(self,amount):
    print(f"Paid Rs.{amount} using Credit Card.")
class UpiPayment(Payment):
  def pay(self,amount):
    print(f"Paid Rs.{amount} using UPI.")
payments=[CreditCardPayment(),UpiPayment()]
for method in payments:
  method.pay(15000)
try:
  p=Payment()
except TypeError as e:
  print(f"\nError:{e}")
```

```python
# 7. Abstract Class with concrete +Abstract Methods
from abc import ABC,abstractmethod
class Appliance(ABC):
  def __init__(self,brand):
    self.brand=brand
  def power_info(self):
    print(f"{self.brand} appliance is plugged in and ready.")
  @abstractmethod
  def turn_on(self):
    pass
class WashingMachine(Appliance):
  def turn_on(self):
    print(f"{self.brand} washing machine starts spinning.")
class Refrigerator(Appliance):
  def turn_on(self):
    print(f"{self.brand}refrigerator starts cooling.")
appliances=[WashingMachine("LG"),Refrigerator("Samsung")]
for a in appliances:
  a.power_info()
  a.turn_on()
  print()
```

```python
# 8.Abstraction with Multiple Abstract Methods(Interface-style)
from abc import ABC,abstractmethod
class Notification(ABC):
  @abstractmethod
  def send(self,message):
    pass
  @abstractmethod
  def get_channel(self):
    pass
class EmailNotification(Notification):
  def send(self,message):
    print(f"[Email] Sending:{message}")
  def get_channel(self):
    return "Email"
class SmsNotification(Notification):
  def send(self,message):
    print(f"[SMS] Sending:{message}")
  def get_channel(self):
    return"SMS"
class PushNotification(Notification):
  def send(self,message):
    print(f"[Push] Sending:{message}")
  def get_channel(self):
    return "Push Notification"
channels=[EmailNotification(),SmsNotification(),PushNotification()]
for channel in channels:
  print(f"Channel:{channel.get_channel()}")
  channel.send("Your order has been shipped")
```

```python
# 9. Combining Multiple Inheritance and Abstraction
from abc import ABC,abstractmethod
class Drivable(ABC):
  def drive(self):
    pass
class Floatable(ABC):
  @abstractmethod
  def float_on_water(self):
    pass
class AmphibiousVehicle(Drivable,Floatable):
  def __init__(self,name):
    self.name=name
  def drive(self):
    print(f"{self.name} is drivable on the road.")
  def float_on_water(self):
    print(f"{self.name} is floating on water.")
vehicle=AmphibiousVehicle("Aqua Cruiser")
vehicle.drive()
vehicle.float_on_water()

```

```python
# 10. Diamond Inheritance and Method Resolution Order (MRO)
class Device:
  def info(self):
    print("This is an electronic device.")
class Chargeable(Device):
  def charge(self):
    print("Charging the device....")
class Connectable(Device):
  def connect(self):
    print("Connecting to Bluetooth...")
class SmartWatch(Chargeable,Connectable):
  def show_features(self):
    self.info()
    self.charge()
    self.connect()
watch=SmartWatch()
watch.show_features()
print("\nMethod Resolution Order:")
for cls in SmartWatch.__mro__:
  print(cls.__name__)
```

```python
# 11.Abstract Base Class Enforcing Structure
from abc import ABC,abstractmethod
class DatabaseConnector(ABC):
  @abstractmethod
  def connect(self):
    pass
  @abstractmethod
  def disconnect(self):
    pass
  def query(self,sql):
    self.connect()
    print(f"Executing query:{sql}")
    self.disconnect()
class MySQLConnector(DatabaseConnector):
  def connect(self):
    print("Connected to MYSQL database.")
  def disconnect(self):
    print("Disconnected from MYSQL database.")
class MongoDBConnector(DatabaseConnector):
  def connect(self):
    print("Connected to MongoDB database.")
  def disconnected(self):
    print("Disconnected from MangoDB database.")
db1=MySQLConnector()
db1.query("SELECT * FROM students")
print()
db2=MongoDBConnector()
db2.query('{"find":"student"}')

```

```python
# 13. hierarchial Inheritance with Abstraction(Payroll System)
from abc import ABC,abstractmethod
class Employee(ABC):
  def __init__(self,name):
    self.name=name
  @abstractmethod
  def calculate_pay(self):
    pass
  def display_pay(self):
    print(f"{self.name} earns Rs.{self.calculate_pay():.2f}")
class SalariedEmployee(Employee):
  def __init__(self,name,monthly_salary):
    super().__init__(name)
    self.monthly_salary=monthly_salary
  def calculate_pay(self):
    return self.monthly_salary
class HourlyEmployee(Employee):
  def __init__(self,name,hours_worked,hourly_rate):
    super().__init__(name)
    self.hours_worked=hours_worked
    self.hourly_rate=hourly_rate
  def calculate_pay(self):
    return self.hours_worked*self.hourly_rate
class CommissionEmployee(Employee):
  def __init__(self,name,sales,commission_rate):
    super().__init__(name)
    self.sales=sales
    self.commission_rate=commission_rate
  def calculate_pay(self):
    return self.sales*self.commission_rate
employees=[
    SalariedEmployee("Neha Gupta",45000),
    HourlyEmployee("Arjun Mehta",160,250),
    CommissionEmployee("Kabir Malhotra",200000,0.05)
]
for emp in employees:
  emp.display_pay()
```

```python
# 14 . Abstraction for interest Calculation
from abc import ABC,abstractmethod
class Account(ABC):
  def __init__(self,holder_name,balance):
    self.holder_name=holder_name
    self._balance=balance
  @abstractmethod
  def calculate_interest(self):
    pass
  def apply_interest(self):
    interest=self.calculate_interest()
    self._balance+=interest
    print(f"{self.holder_name}:Interest Rs.{interest:.2f} added. New Balance : Rs.{self._balance:.2f}")
class SavingsAccount(Account):
  RATE=0.04
  def calculate_interest(self):
    return self._balance*self.RATE
class FixedDepositAccount(Account):
  RATE=0.07
  def calculate_interest(self):
    return self._balance*self.RATE
accounts=[SavingsAccount("Meera Iyer",50000),
          FixedDepositAccount("Rohan Das",100000)]
for acc in accounts:
  acc.apply_interest()
```

```python
# 15. Caapstone-E-commerce Order Processing
from abc import ABC,abstractmethod
class Order(ABC):
  def __init__(self,order_id,customer,amount):
    self.order_id=order_id
    self.customer=customer
    self.amount=amount
  @abstractmethod
  def calculate_shipping(self):
    pass
  @abstractmethod
  def deliver(self):
    pass
  def generate_invoice(self):
    total=self.amount+self.calculate_shipping()
    print(f"\nInvoice for Order #{self.order_id}({self.customer})")
    print(f" Item Amount : Rs.{self.amount:.2f}")
    print(f"  Shipping  :Rs.{self.calculate_shipping():.2f}")
    print(f"  Total    :Rs.{total:.2f}")
class DigitalOrder(Order):
  def calculate_shipping(self):
    return 0.0
  def deliver(self):
    print(" Delivery  :Sent via download link.")
class PhysicalOrder(Order):
  def __init__(self,order_id,customer,amount,weight_kg):
    super().__init__(order_id,customer,amount)
    self.weight_kg=weight_kg
  def calculate_shipping(self):
    return self.weight_kg*40
  def deliver(self):
    print("Delivery   :Dispatched via courier partner.")
orders=[
    DigitalOrder("ORD1001","Sneha Kapoor",999),
    PhysicalOrder("ORD1002","Aditiya Joshi",2500,weight_kg=3)
    ]
for order in orders:
  order.generate_invoice()

```
