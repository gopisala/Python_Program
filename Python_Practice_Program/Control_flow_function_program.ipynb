```python
#PYTHON INTERVIEW PROGRAMMING-CONTROL FLOW & FUNCTION
# 1.Write a function to find the first number divisible by both 7 and 11
def first_common_multiple(strat,end):
  for n in range(strat,end+1):
    if n % 7==0 and n%11==0:
      return n
  return None
print(first_common_multiple(1,1000))
```

```python
# 2.Search a List with break and continue
def search_valid_target(values,target):
  for value in values:
    if value is None:
      continue
    if value < 0:
      continue
    if value==target:
      return True
    if value > target:
      break
    return False
data=[-4,None,2,5,8,13,21]
print(search_valid_target(data,13))
print(search_valid_target(data,7))
```

```python
# 3. Print Prime Numbers Using break
def primes_up_to(limit):
 result=[]
 for n in range(2,limit+1):
  is_prime=True
  for divisor in range(2,int(n**0.5)+1):
   if n% divisor==0:
    is_prime=False
    break
  if is_prime:
   result.append(n)
 return result
print(primes_up_to(50))
```

```python
# 4.Skip Multiples and Stop at  a Threshold
def process_numbers(limit,stop_at):
  total=0
  for n in range(1,limit+1):
    if n%3==0:
      continue
    if n>stop_at:
      break
    total+=n
  return total
print(process_numbers(100,20))
```

```python
# 5.Validate Password Rules with for+if/else
def validate_password(password):
 if len(password)<8:
  return False,"Minimum 8 characters required"
 has_upper=has_lower=has_digit=False
 for ch in password:
  if ch.isupper():
    has_upper=True
  elif ch.islower():
    has_lower=True
  elif ch.isdigit():
    has_digit=True
 if has_upper and has_lower and has_digit:
  return True,"Valid password"
 return False,"Need upperrcase,lowercase and digit"
print(validate_password("DataScience9"))
```

```python
# 7. Find the second Largest Distinct Number
def second_largest(values):
  largest=second=None
  for x in values:
    if largest is None or x > largest:
      if x!=largest:
        second=largest
        largest=x
    elif x!=largest and (second is None or x > second):
      second=x
  return second
print(second_largest([10,5,10,8,7,8]))
```
