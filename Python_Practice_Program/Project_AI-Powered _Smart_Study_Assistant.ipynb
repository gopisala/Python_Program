````python
#ENVIRONMENT SETUP & API CONFIGURATION
import requests
import json
from getpass import getpass

# ==============================
# OPENROUTER API SETUP
# ==============================

API_KEY = getpass("Enter your OpenRouter API Key: ")

MODEL = "openai/gpt-4o-mini"

headers = {
    "Authorization": f"Bearer {API_KEY}",
    "Content-Type": "application/json"
}

print("OpenRouter API connected successfully!")


# ==============================
# FUNCTION TO CALL OPENROUTER
# ==============================

def ask_ai(prompt):

    data = {
        "model": MODEL,
        "messages": [
            {
                "role": "user",
                "content": prompt
            }
        ]
    }

    response = requests.post(
        "https://openrouter.ai/api/v1/chat/completions",
        headers=headers,
        json=data
    )

    if response.status_code != 200:
        raise Exception(
            f"API Error {response.status_code}: {response.text}"
        )

    return response.json()["choices"][0]["message"]["content"]


# ==============================
# GET STUDY NOTES
# ==============================

initial_notes_input = input("""
Paste your study notes here:

""")

if not initial_notes_input.strip():
    raise ValueError("Please enter some study notes.")

print("\nNotes received successfully!")
print("Word count:", len(initial_notes_input.split()))

notes = initial_notes_input


# ==============================
# AI SUMMARY
# ==============================

prompt = f"""
You are an AI-powered study assistant.

Summarize the following study notes.

Requirements:
- Use clear bullet points.
- Identify important concepts.
- Include definitions.
- Include important facts.
- Keep the summary concise.
- Use ONLY the information in the notes.

STUDY NOTES:
{notes}
"""

summary = ask_ai(prompt)

print("\n========== AI SUMMARY ==========\n")
print(summary)


# ==============================
# AI QUIZ
# ==============================

quiz_prompt = f"""
Create 5 multiple-choice questions from the study notes below.

Return ONLY valid JSON using this format:

[
  {{
    "question": "Question",
    "options": {{
      "A": "Option A",
      "B": "Option B",
      "C": "Option C",
      "D": "Option D"
    }},
    "answer": "A",
    "explanation": "Explanation"
  }}
]

Rules:
- Exactly 5 questions.
- Four options for every question.
- Only one correct answer.
- Questions must be based ONLY on the notes.

STUDY NOTES:
{notes}
"""

raw_quiz = ask_ai(quiz_prompt).strip()

# Remove markdown formatting if AI adds it
raw_quiz = raw_quiz.replace("```json", "")
raw_quiz = raw_quiz.replace("```", "")
raw_quiz = raw_quiz.strip()

quiz = json.loads(raw_quiz)

print("\n========== AI QUIZ ==========\n")

for i, q in enumerate(quiz, 1):

    print(f"Q{i}. {q['question']}")

    for letter, option in q["options"].items():
        print(f"   {letter}) {option}")

    print(f"Answer: {q['answer']}")
    print(f"Explanation: {q['explanation']}")

    print("\n======================================")


# ==============================
# ASK QUESTIONS
# ==============================

print("\n======================================")
print("       AI STUDY QUESTION ANSWER")
print("======================================")

while True:

    question = input("\nAsk a question (type exit to stop): ")

    if question.lower().strip() == "exit":
        print("Study session ended.")
        break

    question_prompt = f"""
You are a study assistant.

Answer the question using ONLY the study notes below.

If the answer is not available in the notes, say:
"The answer is not available in the provided study notes."

STUDY NOTES:
{notes}

QUESTION:
{question}
"""

    answer = ask_ai(question_prompt)

    print("\nAI Answer:")
    print(answer)


# ==============================
# CREATE REPORT
# ==============================

report = []

report.append("=" * 60)
report.append("AI-POWERED SMART STUDY ASSISTANT")
report.append("=" * 60)

report.append("\n\nORIGINAL STUDY NOTES")
report.append("-" * 60)
report.append(notes)

report.append("\n\nAI-GENERATED SUMMARY")
report.append("-" * 60)
report.append(summary)

report.append("\n\nSELF-ASSESSMENT QUIZ")
report.append("-" * 60)

for i, q in enumerate(quiz, 1):

    report.append(f"\nQ{i}. {q['question']}")

    for letter, option in q["options"].items():
        report.append(f"{letter}) {option}")

    report.append(f"Correct Answer: {q['answer']}")
    report.append(f"Explanation: {q['explanation']}")


report.append("\n\n" + "=" * 60)
report.append("END OF REPORT")
report.append("=" * 60)

final_report = "\n".join(report)


# ==============================
# SAVE REPORT
# ==============================

with open("study_report.txt", "w", encoding="utf-8") as file:
    file.write(final_report)

print("\nStudy report created successfully!")


# ==============================
# DOWNLOAD IN GOOGLE COLAB
# ==============================

from google.colab import files

files.download("study_report.txt")
````
