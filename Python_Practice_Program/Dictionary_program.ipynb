```python
#PYTHON - DICTIONARY DATA STRUTURE

#Creating dictionary
d={10:"Ramesh",20:"Arjun",30:"Daniel"}
print(d)
```

```python
#Empty dictionary
#creating empty dictionary
d={}
print(d)
print(type(d))
```

```python
#Accessing dictioonary values by using
d={10:"Ramesh",20:"Arujun",30:"Daniel"}
print(d[10])
print(d[20])
print(d[30])
```

```python
#Creatting empty dictionary and adding elements
d={}
d[10]="Ramesh"
d[20]="Arjun"
d[30]="Daniel"
print(d)
```

```python
#Accessing key amd value from dictionary using for loop
d={10:"Ramesh",20:"Arjun",30:"Daniel"}
for k in d:
  print(k,d[k])
```

```python
#UPDATE DICTIONARY
#case 1
#Adding key-value pair to dictionary
d={10:"Ramesh",20:"Arjun",30:"Daniel"}
print("Old dictionary:",d)
d[99]="John"
print("Added key-value 99:John pair to dictionary:",d)
```

```python
#case 2
#Updating key-value pair in dictionary
d={10:'Ramesh',20:'Arjun',30:'Daniel'}
print("Old dictionary:",d)
d[30]='chandhu'
print("Updated dictionay 3:Chandhu:",d)
```

```python
#Removing or deleting elements from dictionary

#Deleting key in dictionary
d={10:"Ramesh",20:"Arjun",30:"Daniel"}
print("Before deleting key from dictionary:",d)
del d[10]
print("After deleting key from dictionary:",d)
```

```python
#Delete key in dictionary
d={10:"Ramesh",20:"Arjun",30:"Daniel"}
print(d)
del d
print(d)
```

```python
#Len(p) function
# Finding length of dictionary
d={100:"Ramesh",200:"Arjun"}
print("Length of dictionary is:",len(d))
```

```python
#METHODS IN DICT CLASS DAT STRUCTURE
#Printing dict data structure methods by using dir(dict)function
print(dir(dict))
```

```python
#Clear() method
#Removing dictionary object by using clear()method
d={10:"Ramesh",20:"Arjusn",30:"Daniel"}
d.clear()
print("After cleared entries in dictionary:",d)
```

```python
#Key() method
d={100:"Ramesh",200:"Arjun",300:"Daniel"}
print(d)
print(d.keys())
```

```python
# Values()
d={100:"Raja",200:"Gopika",300:"Chlm"}
print(d)
print(d.values())
```

```python
#Items()

d={10:"Ramesh",20:"Arjun",30:"Daniel"}
print(d)
print(d.items())
```

```python
#items
#Accessing key and value from dictionary using items() method
d={10:"Ramesd",20:"Arjun",30:"Daniel"}
for k,v in d.items():
  print(k,v)
```

```python
#Dictionary comprehension
squares={a:a*a for a in range(1,6)}
print(squares)
```
