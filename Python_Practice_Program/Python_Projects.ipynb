````python
# PROJECT 2 (Bank Management System)
#Hint 1 BankAccount Class Skeleton
import random
import datetime
class BankAccount:
  def __init__(self,name,pin,initial_deposit=500):
   self.acc_no   =random.randint(100000,999999)
   self.name   =name
   self.__pin  =pin
   self.balance   =initial_deposit
   self.transaction=[]
   self.locked   =False
   self.wrong_attempts=0
   self._log('CR',initial_deposit,'Account opened')
  def _log(self,t_type,amount,note=""):
    self.transaction.append({
        'date'  :datetime.datetime.now().strftime('%d-%m-%Y %H:%M'),
        'type'  :t_type,
        'amount':amount,
        'note'   :note,
        'balance':self.balance
    })
# Hint 2. PIN Verification & Lockout
def verify_pin(self,pin):
  if self.locked:
    print('Account is locked. Contact bank.')
    return True
  else:
    self.wrong_attempts+=1
    remaining=3-self.wrong_attempts
    if remaining<=0:
      self.locked=True
      print('Account Locked after 3 wrong attempts!')
    else:
      print(f'Wrong PIN.{remaining} attempts(s) left.')
    return False
def change_pin(self):
  old=input("Enter current PIN:")
  if not self.verify_pin(old):return
  new1=input('Enteer neww PIN (4 digits):')
  new2=input('Confirm new PIN:')
  if new1 == new2 and len(new1) == 4 and new1.isdigit():
    self.__pin=new1
    print('PIN changed sucessfully!')
  else:
    print('PINs do not match or invalid format.')
# 3. Deposit() and withdraw()
def deposit(self,amount):
  if amount<1:
    print('minimum desposit is Rs.1.')
    return False
  self.balance+=amount
  self._log('CR',amount,'Deposit')
  print(f'Rs.{amount} deposited.New balance:Rs.{self.balance:.2f}')
  return True
def withdraw(self,amount):
  MIN_BALANCE=500
  if amount<1:
    print('Minimum withdrawal is Rs.1.')
    return False
  if self.balance-amount<MIN_BALANCE:
    print(f'Insufficient funds.Min balance Rs.{MIN_BALANCE} required.')
    return False
  self.balance-=amount
  self._log('DR',amount,'Withdrawal')
  print(f'Rs.{amount} withdrawn.New Balance: Rs.{self.balance:.2f}')
  return True
# hint 4 transfer() Between Accounts
def transfer(self,target_account,amount):
  print(f'Transferring Rs.{amount} to {target_account.name}...')
  if self.withdraw(amount):
    target_account.deposit(amount)
    self._log('DR',amount,f'Transfer to {target_account.acc_no}')
    target_account._log('CR',amount,f'Transfer from {self.acc_no}')
    print('Transfer successful')
  else:
    print('Transfer failed.')
#5 mini_statement()
def mini_statement(self):
  print(f'\n---Mini Statement:{self.name} ({self.acc_no}) ---')
  last5=self.transaction[-5]
  print(f"{'Date':<18}{'Type':<6}{'Amount':>10}{'Balance':>12}")
  print('-'*48)
  for t in last5:
    print(f"{t['date']:<18}{t['type']:<6}"
     f"{t['amount']:>10.2f}{t['balance']:>12.2f}")
def check_balance(self):
  print(f'Account Holder :{self.name}')
  print(f'Account Number : {self.acc_no}')
  print(f'Balance   :Rs.{self.balance:.2f}')

def apply_interest(self,rate=4.0):
  interest=set.balance*rate/100
  self.balance+=interest
  self._log('CR',interest,f'{rate}% annual interest')
  print(f'Interest Rs.{interest:.2f} added.New balnce:Rs.{self.balnce:.2f}')
# main program & Menu Loop
accounts={}
def create_account():
  name=input('Full Name:').strip()
  pin = input('Set 4_digit PIN:')
  if not (pin.isdigit() and len(pin)==4):
    print('PIN must be exactly 4 digits.');return
  dep=float(input('Initial deposit(min Rs.500):'))
  if dep<500:
    print('Minimum opening deposit is Rs.500');return
  acc=BankAccount(name,pin,dep)
  accounts[acc.acc_no]=acc
  print(f'Account created! Your account number:{acc.acc_no}')
def main_menu():
  while True:
    print('\n===BANK MANAGEMENT SYSTEM ===')
    print('1.Create Account 2.Login 3.Exit')
    ch=input('Choice.')
    if ch=='1':create_account()
    elif ch=='2':
      acc_no=int(input('Account Number:'))
      if acc_no not in accounts:
        print('Account not foun d.');continue
      acc=accounts[acc_no]
      pin=input('PIN:')
      if not acc.verify_pin(pin):continue
      account_menu(acc)
    elif ch=='3':print('Thank you!');break
def account_menu(acc):
  ops={
      '1':lambda:acc.check_balance(),
      '2':lambda:acc.deposit(float(input('Amount:'))),
      '3':lambda:acc.withdraw(float(input('Amount:'))),
      '4':lambda:acc.mini_statement(),
      '5':lambda:acc.apply_interest(),
      '6':lambda:acc.change_pin(),
  }
  while True:
    print('\n1.Balance 2.Deposit 3.Withdraw 4.Statement 5.Interest 6.PIN 7.Exit')
    ch=input('Choice:')
    if ch=='7':break
    elif ch in ops:ops[ch]()
if __name__=='__main__':
  main_menu()








````

````python
# PYTHON PROJECT-STUDENT MANAGEMENT SYSTEM
# Hint 1 Data Structure & Grade Helper
students=[]
def get_grade(average):
  if average >= 90:return 'A+'
  elif average >=80:return 'A'
  elif average >=70: return 'B'
  elif average >=60:return 'c'
  elif average >=50: return 'D'
  else:return 'F'
# hint 2 add_student

def add_student():
  roll=int(input('Enter Roll Number:'))
  name=input('Enter Name:').strip()
  age=int(input('Enter Age:'))
  marks=[]
  subjects=['Tamil','English','Maths','Science','Social']
  for sub in subjects:
    while True:
     try:
      m=float(input(f'  {sub} marks (0-100):'))
      if 0<=m<=100:
        marks.append(m)
        break
      else:
          print('Marks must be between 0 and 100.')
     except ValueError:
      print('Please enter a number.')
  student={'roll':roll,'name':name,'age':age,'marks':marks}
  students.append(student)
  print(f'Student {name} added successfully!')

# Hint 3 view_all_Students()
def view_all_students():
  if not students:
    print('No student records found.')
    return
  print(f"{'Roll':<6}{'Name':<18}{'Age':<5}{'Total':<8}{'Avg':<8}{'Grade'}")
  print('-'*55)
  for s in students:
    total=sum(s['marks'])
    avg=total/len(s['marks'])
    grade=get_grade(avg)
    print(f"{s['roll']:<6}{s['name']:<18}{s['age']:<5}"
        f"{total:<8.1f}{avg:<8.1f}{grade}")
# Hint 4 search_student()  & find helper
def find_student(roll):
  for s in students:
    if s['roll']==roll:
      return s
  return None
def search_student():
  roll=int(input('Enter Roll Number to search:'))
  s=find_student(roll)
  if s:
    total=sum(s['marks'])
    avg=total/len(s['marks'])
    print(f"Name:{s['name']}")
    print(f"Age:{s['age']}")
    print(f"Marks:{s['marks']}")
    print(f"Total:{total} Avg:{avg} Grade:{get_grade(avg)}")
  else:
    print('Student not found.')
# Hint 5 updated_student()
def update_student():
  roll=int(input('Enter Roll Number to update:'))
  s=find_student(roll)
  if not s:
    print('Student not found.')
    return
  print('1.Update Name 2. Update Age 3.Update Marks')
  choice=input('Choose field:')
  if choice=='1':
    s['name']=input('New name:').strip()
  elif choice=='2':
    s['age']=int(input('New age:'))
  elif choice=='3':
    print('Record updated!')
# Hint delete _student()
def delete_student():
  roll=input(input('Enter Roll Number to delete:'))
  s=find_student(roll)
  if not s:
    print('Student not found.')
    return
  confirm=input(f"Delete {s['name']}? (yes/no):")
  if confirm.lower()=='yes':
    students.remove(s)
    print('Student deleted.')
  else:
    print('Deletion cancelled.')
# 7.show_statistics()
def show_statistics():
  if not students:
    print('No data.')
    return
  avgs=[sum(s['marks'])/len(s['marks']) for s in students]
  best=students[avgs.index(max(avgs))]
  worst=students[avgs.index(min(avgs))]
  print(f'Total Student:{len(students)}')
  print(f'Class Average:{sum(avgs)/len(avgs):.2f}')
  print(f'Top Scorer:{best["name"]}({max(avgs):.1f})')
  print(f'Low Scorer:{worst["name"]}({min(avgs):.1f})')
# 8. main_menu()Entry Point
def main_menu():
  menu={
    '1':add_student,
    '2':view_all_students,
    '3':search_student,
    '4':update_student,
    '5':delete_student,
    '6':show_statistics,
  }
  while True:
    print('\n==STUDENT MANAGEMENT SYSTEM===')
    print('1.Add 2.View 3.Search 4.Update 5.Delete 6.Stats 7.Exit')
    choice=input('Enter chocie:').strip()
    if choice=='7':
      print('Goodbye!')
      break
    elif choice in menu:
      menu[choice]()
    else:
      print('Invalid choice.Try again.')
if __name__=='__main__':
  main_menu()














````

````python
#AI-POWERED SMART STUDY(USING OPENROUTER KEY)
import requests
import json
from getpass import getpass

# ==============================
# OPENROUTER API SETUP
# ==============================

API_KEY = getpass("Enter your OpenRouter API Key: ")

MODEL = "openai/gpt-4o-mini"

headers = {
    "Authorization": f"Bearer {API_KEY}",
    "Content-Type": "application/json"
}

print("OpenRouter API connected successfully!")


# ==============================
# FUNCTION TO CALL OPENROUTER
# ==============================

def ask_ai(prompt):

    data = {
        "model": MODEL,
        "messages": [
            {
                "role": "user",
                "content": prompt
            }
        ]
    }

    response = requests.post(
        "https://openrouter.ai/api/v1/chat/completions",
        headers=headers,
        json=data
    )

    if response.status_code != 200:
        raise Exception(
            f"API Error {response.status_code}: {response.text}"
        )

    return response.json()["choices"][0]["message"]["content"]


# ==============================
# GET STUDY NOTES
# ==============================

initial_notes_input = input("""
Paste your study notes here:

""")

if not initial_notes_input.strip():
    raise ValueError("Please enter some study notes.")

print("\nNotes received successfully!")
print("Word count:", len(initial_notes_input.split()))

notes = initial_notes_input


# ==============================
# AI SUMMARY
# ==============================

prompt = f"""
You are an AI-powered study assistant.

Summarize the following study notes.

Requirements:
- Use clear bullet points.
- Identify important concepts.
- Include definitions.
- Include important facts.
- Keep the summary concise.
- Use ONLY the information in the notes.

STUDY NOTES:
{notes}
"""

summary = ask_ai(prompt)

print("\n========== AI SUMMARY ==========\n")
print(summary)


# ==============================
# AI QUIZ
# ==============================

quiz_prompt = f"""
Create 5 multiple-choice questions from the study notes below.

Return ONLY valid JSON using this format:

[
  {{
    "question": "Question",
    "options": {{
      "A": "Option A",
      "B": "Option B",
      "C": "Option C",
      "D": "Option D"
    }},
    "answer": "A",
    "explanation": "Explanation"
  }}
]

Rules:
- Exactly 5 questions.
- Four options for every question.
- Only one correct answer.
- Questions must be based ONLY on the notes.

STUDY NOTES:
{notes}
"""

raw_quiz = ask_ai(quiz_prompt).strip()

# Remove markdown formatting if AI adds it
raw_quiz = raw_quiz.replace("```json", "")
raw_quiz = raw_quiz.replace("```", "")
raw_quiz = raw_quiz.strip()

quiz = json.loads(raw_quiz)

print("\n========== AI QUIZ ==========\n")

for i, q in enumerate(quiz, 1):

    print(f"Q{i}. {q['question']}")

    for letter, option in q["options"].items():
        print(f"   {letter}) {option}")

    print(f"Answer: {q['answer']}")
    print(f"Explanation: {q['explanation']}")

    print("\n======================================")


# ==============================
# ASK QUESTIONS
# ==============================

print("\n======================================")
print("       AI STUDY QUESTION ANSWER")
print("======================================")

while True:

    question = input("\nAsk a question (type exit to stop): ")

    if question.lower().strip() == "exit":
        print("Study session ended.")
        break

    question_prompt = f"""
You are a study assistant.

Answer the question using ONLY the study notes below.

If the answer is not available in the notes, say:
"The answer is not available in the provided study notes."

STUDY NOTES:
{notes}

QUESTION:
{question}
"""

    answer = ask_ai(question_prompt)

    print("\nAI Answer:")
    print(answer)


# ==============================
# CREATE REPORT
# ==============================

report = []

report.append("=" * 60)
report.append("AI-POWERED SMART STUDY ASSISTANT")
report.append("=" * 60)

report.append("\n\nORIGINAL STUDY NOTES")
report.append("-" * 60)
report.append(notes)

report.append("\n\nAI-GENERATED SUMMARY")
report.append("-" * 60)
report.append(summary)

report.append("\n\nSELF-ASSESSMENT QUIZ")
report.append("-" * 60)

for i, q in enumerate(quiz, 1):

    report.append(f"\nQ{i}. {q['question']}")

    for letter, option in q["options"].items():
        report.append(f"{letter}) {option}")

    report.append(f"Correct Answer: {q['answer']}")
    report.append(f"Explanation: {q['explanation']}")


report.append("\n\n" + "=" * 60)
report.append("END OF REPORT")
report.append("=" * 60)

final_report = "\n".join(report)


# ==============================
# SAVE REPORT
# ==============================

with open("study_report.txt", "w", encoding="utf-8") as file:
    file.write(final_report)

print("\nStudy report created successfully!")


# ==============================
# DOWNLOAD IN GOOGLE COLAB
# ==============================

from google.colab import files

files.download("study_report.txt")
````
