```python
#PYTHON-TUPLE DATA STRUTURE
# tuple having same type of objects
employee_ids=(10,20,30,40,50)
print("same type of objects:",employee_ids)
print(type(employee_ids))
```

```python
#A single value with tuple syntax but its not tuple
number=(9)
print(number)
print(type(number))
```

```python
#A single value with tuple syntax but its not tuple
name=("Daniel")
print(name)
print(type(name))
```

```python
#Tuple single value ends with comma separator its tuple
name=("Daniel",)
print(name)
print(type(name))
```

```python
#Parenthesis symblos is optional while creating tuple
emp_ids=10,20,30,40
print(emp_ids)
```

```python
#Empty tuple
emp_id=()
print(emp_id)
print(type(emp_id))
```

```python
#Tuple example
emp_id=(11,212,13)
std_id=120,130,140
print(emp_id)
print(std_id)
```

```python
#Creating tuple by using tuple funtion
a=[11,22,33]
t=tuple(a)
print(t)
```

```python
#Accessing tuple by using index
t=(10,20,30,40,50,60)
print(t[0])
print(t[-1])
```

```python
#Accessing tuple by using slice
t=(10,20,30,40,50,60)
print(t[2:5])
print(t[2:100])
print(t[::2])
```

```python
#CONCATENTENATION OPERATOR ON TUPLE
#cONCATENATION operator on tuple
t1=(10,20,30)
t2=(40,50,60)
t3=t1+t2
print(t3)

```

```python
#MULTIPLICATION OPERATOR(*)

#rEPETITON OPERATOR ON TUPLE

t1=(10,20,30)
t2=t1*3
print(t2)
```
