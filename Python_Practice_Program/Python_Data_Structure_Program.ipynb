```python
# 26. iterate over keys,values and items [dictionary]
marks={"Math":90,"science":85,"English":78,"History":92}
print("------Subject Marks---")
for subject,score in marks.items():
  grade="Pass" if score>=80 else "Needs Work"
  print(f"{subject:10s}:{score} ({grade})")
```

```python
# 27. Count word Frequence in a Sentence  [Dictionary]
sentence='the cat sat on the mat the cat sat'
freq={}
for word in sentence.split():
  freq[word]=freq.get(word,0)+1
  print("Word Frequrncies")
  for word ,count in sorted(freq.items()):
    print(f'{word}:{count}')
```

```python
#28 Merge two Dictionaries  [DICTIONARY]
dict1={'a':1,'b':2,'c':3}
dict2={'c':99,'d':4,'6':5}

#dict2 values overwrite dict1 for duplication keys
merged={** dic1,**dict2}
print('Dict1:',dict1)
print('Dict2:',dict2)
print('Merged:',merged)
```

```python
# Sort a Dictionary by Value  [Dictionary]
scores={'Alice':88,'bob':95,'Charlie':72,'Diana':91}
sorted_scores=dict(sorted(scores.items(),
                          key=lambda x:x[1],
                          reverse=True))
print('Ranked Scores:')
for rank,(name,score) in enumerate(sorted_scores.items(),1):
  print(f'{rank},{name}:{score}')
```
